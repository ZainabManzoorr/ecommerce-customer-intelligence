```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
customer_data = pd.read_csv("customer_features.csv")
```

```python
fig,axs = plt.subplots(1,3,figsize =(18,5))
sns.histplot(customer_data['Recency'],bins=30,ax=axs[0],color='skyblue')
axs[0].set_title("Recency Distribution")

sns.histplot(customer_data['Frequency'], bins=30, ax=axs[1], color='lightgreen')
axs[1].set_title("Frequency Distribution")

sns.histplot(customer_data['Monetary'], bins=30, ax=axs[2], color='salmon')
axs[2].set_title("Monetary Distribution")
plt.show()
```

```python
sns.histplot(customer_data['CLV'], bins=50, color='purple')
plt.title("Customer Lifetime Value Distribution")
plt.xlabel("CLV")
plt.ylabel("Count of Customers")
plt.show()
```

```python
sns.scatterplot(x='AverageBasketSize', y='CLV', data=customer_data)
plt.title("Average Basket Size vs Customer Lifetime Value")
plt.show()
```
