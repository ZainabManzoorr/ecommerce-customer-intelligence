```python
from sklearn.cluster import KMeans
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
customer_data = pd.read_csv("customer_features.csv")
rfm_features = customer_data[['Recency','Frequency','Monetary']]
```

```python
kmeans = KMeans(n_clusters=4,random_state=42)
customer_data['Segment'] = kmeans.fit_predict(rfm_features)
```

```python
sns.scatterplot(x='Recency',y='Frequency',hue='Segment',data=customer_data,palette='Set2')
plt.title("Customer Segments (RFM)")
plt.show()
```

```python
import datetime as dt
transactions = pd.read_csv("Transactions.csv")
monthly_country = transactions.groupby([transactions['InvoiceDate'].dt.to_period('M'), 'Country'])['Revenue'].sum().unstack()
plt.figure(figsize=(15,6))
sns.heatmap(monthly_country.fillna(0), cmap='YlGnBu')
plt.title("Monthly Revenue by Country")
plt.show()
```
